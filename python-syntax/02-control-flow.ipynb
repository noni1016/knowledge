# 02. 제어문 (if / for / while)

조건문, 반복문, 그리고 반복문에만 있는 `else` 절, 파이썬 3.10+의 `match-case`까지 다룹니다.

## if / elif / else, 삼항 표현식

```python
age = 17
if age >= 20:
    group = "성인"
elif age >= 13:
    group = "청소년"
else:
    group = "어린이"
print(group)

# 삼항(조건부) 표현식: 짧은 분기는 한 줄로
label = "성인" if age >= 20 else "미성년"
print(label)
```

## for 루프: range, enumerate, zip

C 스타일 `for(i=0; i<n; i++)`를 그대로 쓰지 않습니다. 인덱스가 필요하면 `enumerate`, 여러 리스트를 같이 돌려면 `zip`.

```python
names = ["alpha", "beta", "gamma"]

for i, name in enumerate(names, start=1):
    print(i, name)

scores = [90, 85, 70]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

## while 루프와 break / continue / else

파이썬 루프에는 `else` 절이 있습니다 — **break 없이 끝까지 돌았을 때만** 실행됩니다. 흔히 '찾기' 로직에서 유용합니다.

```python
def find_first_negative(nums):
    i = 0
    while i < len(nums):
        if nums[i] < 0:
            print(f"음수 발견: index {i}")
            break
        i += 1
    else:
        print("음수가 없습니다")

find_first_negative([1, 2, -3, 4])
find_first_negative([1, 2, 3, 4])
```

## match-case (Python 3.10+)

C의 switch와 비슷하지만 패턴 매칭이 더 강력합니다 (구조 분해까지 가능).

```python
def describe(value):
    match value:
        case 0:
            return "zero"
        case int() | float() if value < 0:
            return "negative number"
        case [x, y]:
            return f"pair: {x}, {y}"
        case {"type": "sensor", "id": sid}:
            return f"sensor message, id={sid}"
        case _:
            return "something else"

for v in [0, -5, [1, 2], {"type": "sensor", "id": 7}, "hello"]:
    print(v, "->", describe(v))
```

## 헷갈렸던 점 / 더 찾아볼 것

- `for...else`는 이름이 직관적이지 않지만 '정상 종료 시 실행'이라고 외우면 됨
- `match`는 단순 값 비교보다 구조 분해(destructuring)에서 진가를 발휘 — 센서 메시지 dict 파싱 등에 활용 가능
