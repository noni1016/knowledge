# 03. 자료구조 (list / tuple / dict / set) + comprehension

실무에서 가장 많이 쓰는 4가지 내장 컨테이너와, 이들을 한 줄로 만드는 comprehension 문법.

## list: 순서 있음, 가변

슬라이싱 `[start:stop:step]`을 자유자재로 쓸 수 있으면 많은 걸 한 줄로 끝낼 수 있습니다.

```python
nums = [10, 20, 30, 40, 50]

print(nums[1:3])     # [20, 30]
print(nums[:3])      # 처음부터
print(nums[-2:])     # 뒤에서 2개
print(nums[::-1])    # 뒤집기
print(nums[::2])     # 짝수 인덱스만

nums.append(60)
nums.insert(0, 0)
nums.remove(30)       # 값으로 제거
popped = nums.pop()   # 마지막 원소 제거 + 반환
print(nums, popped)
```

## tuple: 순서 있음, 불변

불변이라 딕셔너리 key로 쓸 수 있고, '여러 값을 묶어서 반환'할 때 자주 사용됩니다.

```python
point = (3, 4)
x, y = point  # 언패킹
print(x, y)

def min_max(nums):
    return min(nums), max(nums)  # 튜플로 반환됨

lo, hi = min_max([5, 2, 9, 1])
print(lo, hi)

# 원소가 1개인 튜플은 콤마가 꼭 필요
single = (1,)
print(type(single))
```

## dict: 키-값 매핑 (Python 3.7+부터 삽입 순서 보장)

```python
config = {"lr": 0.001, "batch_size": 32, "epochs": 10}

for key, value in config.items():
    print(key, "=", value)

print(config.get("optimizer", "adam"))  # 없는 키는 기본값
config.setdefault("optimizer", "adam")  # 없을 때만 추가
config.update({"epochs": 20, "seed": 42})
print(config)
```

## set: 순서 없음, 중복 제거, 집합 연산

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # 합집합
print(a & b)   # 교집합
print(a - b)   # 차집합
print(a ^ b)   # 대칭차집합 (한쪽에만 있는 것)

tags = ["python", "mlops", "python", "robot", "mlops"]
print(set(tags))  # 중복 제거
```

## Comprehension: list / dict / set을 한 줄로

`[표현식 for 항목 in 반복가능 if 조건]` 형태. 중첩 for와 if를 조합할 수 있습니다.

```python
squares = [x**2 for x in range(10) if x % 2 == 0]
print(squares)

name_to_len = {name: len(name) for name in ["alpha", "beta", "gamma"]}
print(name_to_len)

unique_lengths = {len(name) for name in ["ab", "cd", "efg", "hi"]}
print(unique_lengths)

# 중첩: 2차원 좌표 평탄화
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [val for row in matrix for val in row]
print(flat)
```

## 헷갈렸던 점 / 더 찾아볼 것

- comprehension 안에 조건이 두 개 들어가면(필터 + 삼항 표현식) 가독성이 급격히 떨어짐 — 그럴 땐 그냥 for문으로 풀어쓰는 게 나음
- `collections.namedtuple` / `dataclass`로 튜플보다 가독성 좋은 구조체 만들기 (06번에서 다룸)
