## 변수 선언과 동적 타이핑

파이썬은 변수 선언 시 타입을 명시하지 않습니다. 변수는 값에 대한 **이름표(참조)** 일 뿐이고, 같은 이름에 다른 타입의 값을 다시 할당할 수 있습니다.

```python
x = 10
print(type(x), x)

x = "이제 문자열"
print(type(x), x)
```

## 숫자 타입: int / float / complex

나눗셈 연산자가 여러 개라 헷갈리기 쉽습니다.

```python
a, b = 7, 2

print("a / b  (실수 나눗셈) =", a / b)
print("a // b (몫, 내림)    =", a // b)
print("a % b  (나머지)      =", a % b)
print("a ** b (거듭제곱)    =", a ** b)

print("-7 // 2 =", -7 // 2)  # 내림이라 음수에서 주의 (-3.5 -> -4)
```

## 문자열과 f-string

포맷팅은 거의 항상 f-string을 씁니다 (`.format()`, `%` 방식은 레거시 코드에서나 봄).

```python
name = "Noni"
score = 95.456

print(f"{name}님의 점수는 {score:.1f}점입니다")  # 소수점 1자리
print(f"{score:>10.2f}")  # 폭 10, 오른쪽 정렬
print(f"{name!r}")  # repr 형태로 출력
```

## bool과 None

`None`은 '값이 없음'을 나타내는 유일한 객체입니다. `== None`이 아니라 `is None`으로 비교하는 게 관례입니다 (성능보다는 의미 차이 때문).

```python
value = None
print(value is None)       # 권장
print(bool(0), bool(""), bool([]), bool(None))  # 모두 False (falsy)
print(bool(1), bool("0"), bool([0]))            # 문자열 "0"은 True! 빈 문자열이 아니므로
```

## type() / isinstance()

타입 체크는 `isinstance`를 쓰는 게 안전합니다. 상속 관계까지 고려해주기 때문입니다.

```python
class Animal: pass
class Dog(Animal): pass

d = Dog()
print(type(d) == Dog)         # True
print(type(d) == Animal)      # False — 정확히 같은 타입만 비교
print(isinstance(d, Animal))  # True — 상속 관계도 인정
```

## 형변환

```python
print(int("42"), float("3.14"), str(3.14), int(3.9))  # int()는 버림(truncate), round()가 아님
print(round(3.567, 2))
```

## 가변(mutable) vs 불변(immutable)

int, float, str, tuple은 불변. list, dict, set은 가변.
`==`는 '값이 같은가', `is`는 '같은 객체(메모리 주소)인가'를 봅니다.

```python
a = [1, 2, 3]
b = a          # 같은 리스트를 가리킴 (복사 아님!)
c = a.copy()   # 새 리스트로 복사

b.append(4)
print("a:", a)  # b를 바꿨는데 a도 바뀜
print("c:", c)  # c는 영향 없음

print(a is b, a is c, a == c)
```

## 헷갈렸던 점 / 더 찾아볼 것

- `is`와 `==`의 차이는 리스트/딕셔너리처럼 가변 객체를 다룰 때 특히 중요 (얕은 복사 vs 깊은 복사, `copy.deepcopy`)
- `int // -2` 같은 음수 내림 나눗셈은 C/C++과 결과가 다를 수 있으니 주의
